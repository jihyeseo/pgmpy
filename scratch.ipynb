```python
from pgmpy.utils import get_example_model

model = get_example_model('asia')

from pgmpy.inference import VariableElimination

infer = VariableElimination(model)

res = infer.query(['dysp'], evidence={'either': 'yes', 'lung': 'no', 'tub': 'no'})
print(res.values)
```

```python
import pandas as pd
from pgmpy.models import BayesianNetwork
from pgmpy.sampling import BayesianModelSampling
from pgmpy.factors.discrete import State
from pgmpy.estimators import MaximumLikelihoodEstimator

# Create simple data where P(X|Y) is 1 when X==Y, and 0 otherwise.
data = {'X': ['1.0', '2.0', '3.0'], 'Y': ['1.0', '2.0','3.0']}
df = pd.DataFrame(data)

# Train the model
model = BayesianNetwork([('X', 'Y')])
estimator = MaximumLikelihoodEstimator(model, df)
cpds = estimator.get_parameters()
model.add_cpds(*cpds)

# Sample from the model
sampler = BayesianModelSampling(model)
evidence = [State('X', '3.0')]
sample = sampler.rejection_sample(evidence=evidence, size=1, show_progress=True)

# This sample will have X != Y
print(sample)
```
